```python
encoding = {
    'Gender': {'Male': 1, 'Female': 0},
    'Married': {'Yes': 1, 'No': 0},
    'Education': {'Graduate': 1, 'Not Graduate': 0},
    'Dependents' : {'0': 0, '1': 1, '2': 2, '4': 4},
    'Self_Employed': {'Yes': 1, 'No': 0},
    'Property_Area': {'Urban': 2, 'Rural': 0, 'Semiurban': 1},
    'Loan_Status': {'Y': 1, 'N': 0}

}
```

```python
from sklearn.model_selection import train_test_split
feature_col = [
    'ApplicantIncome',
    'CoapplicantIncome',
    'LoanAmount',
    'Loan_Amount_Term',
    'Credit_History',
    'Dependents',
    'Gender',
    'Married',
    'Education',
    'Self_Employed',
    'Property_Area'
]
X= df[feature_col]
y = df['Loan_Status']

X_train,X_test,y_train,y_test= train_test_split(X,y, test_size =0.25, random_state=42, stratify = y)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer , StandardScaler

plain_numeric_pipeline= Pipeline(steps = [
    ('scaler', StandardScaler())
])

```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(transformers = [
    ('Plain_numeric_pipeline', plain_numeric_pipeline, feature_col)
])
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
lr_pipeline = Pipeline(steps = [
    ('preprocessor' , preprocessor),
    ('model', LogisticRegression())
])

lr_pipeline.fit(X_train, y_train)
lr_preds = lr_pipeline.predict(X_test)
lr_preds_train = lr_pipeline.predict(X_train)


lr_accuracy_score = accuracy_score(y_test, lr_preds)
lr_accuracy_score_train = accuracy_score(y_train, lr_preds_train)
print("Test Accuracy:", lr_accuracy_score)
print("Train Accuracy:", lr_accuracy_score_train)

```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix(y_test, lr_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay,classification_report

knn_pipeline = Pipeline(steps = [
    ('preprocessor' , preprocessor),
    ('model', KNeighborsClassifier())
])

knn_pipeline.fit(X_train, y_train)
knn_preds = knn_pipeline.predict(X_test)
knn_preds_train = knn_pipeline.predict(X_train)


knn_accuracy_score = accuracy_score(y_test, knn_preds)
knn_accuracy_score_train = accuracy_score(y_train, knn_preds_train)
print("Test Accuracy:", knn_accuracy_score)
print("Train Accuracy:", knn_accuracy_score_train)
print(f"Confusion Matrix : {classification_report(y_test,knn_preds)}")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm = confusion_matrix(y_test, knn_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm)
disp.plot()

plt.title("KNN Regression - Confusion Matrix")
plt.show()
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix,classification_report, ConfusionMatrixDisplay
```

```python
svm_pipeline = Pipeline(steps = [
    ('preprocessor' , preprocessor),
    ('model', SVC())
])

svm_pipeline.fit(X_train, y_train)
svm_preds = svm_pipeline.predict(X_test)
svm_preds_train =svm_pipeline.predict(X_train)


svm_accuracy_score = accuracy_score(y_test, svm_preds)
svm_accuracy_score_train = accuracy_score(y_train, svm_preds_train)
print("Test Accuracy:", svm_accuracy_score)
print("Train Accuracy:", svm_accuracy_score_train)
print(f"Confusion Matrix : {classification_report(y_test,svm_preds)}")
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    'model__n_neighbors': [3, 5, 7, 9, 11],
    'model__weights': ['uniform', 'distance'],
    'model__metric': ['euclidean', 'manhattan']
}

grid_search = GridSearchCV(
    estimator=knn_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)


grid_search.fit(X_train, y_train)
print("Best Parameters:", grid_search.best_params_)
print("Best CV Accuracy:", grid_search.best_score_)
```

```python
from xgboost import XGBClassifier

xgb_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', XGBClassifier(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=2,
        min_child_weight=5,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=5,
        random_state=42,
        eval_metric='logloss'
    ))
])

xgb_pipeline.fit(X_train, y_train)

xgb_preds = xgb_pipeline.predict(X_test)
xgb_preds_train = xgb_pipeline.predict(X_train)

print("XGBoost Accuracy:", accuracy_score(y_test, xgb_preds))
print(f"XGBoost Test Accuracy:" , accuracy_score(y_train,xgb_preds_train))
print(f"Classification report: {classification_report(y_test,xgb_preds)}")
```

```python
param_grid = {
    'model__n_estimators': [100, 200, 300],
    'model__learning_rate': [0.03, 0.05, 0.1],
    'model__max_depth': [2, 3, 4],
    'model__min_child_weight': [1, 3, 5],
    'model__subsample': [0.8, 1.0],
    'model__colsample_bytree': [0.8, 1.0]
}

grid_search = GridSearchCV(
    estimator=xgb_pipeline,
    param_grid=param_grid,
    cv=10,
    scoring='f1_macro',
    n_jobs=-1,
    verbose=1
)
grid_search.fit(X_train, y_train)


```

```python
grid_xgb_preds = grid_search.predict(X_test)
grid_xgb_preds_train = grid_search.predict(X_train)

print("XGBoost with grid search Test Accuracy:",
      accuracy_score(y_test, grid_xgb_preds))

print("XGBoost Train Accuracy:",
      accuracy_score(y_train, grid_xgb_preds_train))

print("Classification Report:")
print(classification_report(y_test, grid_xgb_preds))
```
